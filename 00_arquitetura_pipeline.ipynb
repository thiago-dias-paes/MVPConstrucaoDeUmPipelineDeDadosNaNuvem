# Arquitetura do Pipeline — Medalhão (Bronze / Silver / Gold)

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import FancyBboxPatch

fig, ax = plt.subplots(figsize=(14, 7))
ax.set_xlim(0, 11)
ax.set_ylim(0, 6)
ax.axis("off")

cores = {"Bronze": "#D4453C", "Silver": "#F5A623", "Gold": "#2649B2"}

caixas = {
    "vendas_raw": (1, 5, "Bronze"),
    "vendas_limpo": (1, 3, "Silver"),
    "clusters_sku": (1, 1, "Silver"),
    "fato_vendas_mensal": (6, 3, "Gold"),
    "dim_item_loja": (6, 1, "Gold"),
    "fato_forecast_cluster": (9.3, 2, "Gold"),
    "dimensoes (item, loja, tempo)": (6, 5, "Gold"),
}

pos = {}

for nome in caixas:
    x, y, camada = caixas[nome]
    box = FancyBboxPatch((x - 0.9, y - 0.4), 1.8, 0.8, boxstyle="round,pad=0.05", linewidth=1.5, edgecolor="white", facecolor=cores[camada])
    ax.add_patch(box)
    ax.text(x, y, nome, ha="center", va="center", fontsize=8, color="white", weight="bold")
    pos[nome] = (x, y)

setas = [
    ("vendas_raw", "vendas_limpo"),
    ("vendas_limpo", "clusters_sku"),
    ("vendas_limpo", "fato_vendas_mensal"),
    ("clusters_sku", "dim_item_loja"),
    ("fato_vendas_mensal", "dim_item_loja"),
    ("dim_item_loja", "fato_forecast_cluster"),
    ("dimensoes (item, loja, tempo)", "fato_vendas_mensal"),
]

for origem, destino in setas:
    x1, y1 = pos[origem]
    x2, y2 = pos[destino]
    ax.annotate("", xy=(x2, y2), xytext=(x1, y1), arrowprops=dict(arrowstyle="->", color="white", lw=1.3, shrinkA=30, shrinkB=30))

legenda = [mpatches.Patch(color=cores[l], label=l) for l in cores]
ax.legend(handles=legenda, loc="lower center", ncol=3, frameon=False, labelcolor="white")

fig.patch.set_facecolor("#1a1a1a")
ax.set_facecolor("#1a1a1a")
plt.title("Arquitetura do Pipeline — Medalhão (Bronze / Silver / Gold)", color="white", fontsize=13)
plt.tight_layout()
plt.show()
```
