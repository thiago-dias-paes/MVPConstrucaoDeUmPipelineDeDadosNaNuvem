```python
from pyspark.sql import functions as F

caminho_raw = "/Volumes/bronze/sop_forecast/raw_files/train.csv"

df_bronze = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(caminho_raw)
    .withColumn("_ingestion_ts", F.current_timestamp())
    .withColumn("_fonte", F.lit("kaggle_store_item_demand"))
)

(df_bronze.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("bronze.sop_forecast.vendas_raw"))
```
